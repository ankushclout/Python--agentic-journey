```python
nums = [ 1, 4, 9, 16, 25, 36, 49, 64, 81, 100 ]

x = int (input("enter number"))

idx = 0
while idx < len(nums):
    if nums[idx] == x:
        print("number found", x)
        break
    idx += 1
else:
    print("not found")    
```
